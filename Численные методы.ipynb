```python
M = [[2, -1, 5, 10], [1, 1, -3, -2], [2, 4, 1, 1]]
n = len(M)
i=0
for j in range(i+1, n):
    a_i = M[i][i]
    a_j = M[j][i]
    for k in range(len(M[0])):
        M[j][k] = a_i * M[j][k] - a_j * M[i][k]
for row in M:
    print(row)
```
